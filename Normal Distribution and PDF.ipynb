```python
# Step 1 — Create our supplier lead-time dataset
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
lead_time=np.random.normal(loc=6,scale=1,size=1000)
dataset = pd.DataFrame({"Supplier lead time (days)":lead_time})
dataset.head()
```

```python
#Step 2 — Check Mean and Standard Deviation
```

```python
mean = dataset["Supplier lead time (days)"].mean()
SD = dataset["Supplier lead time (days)"].std()
print ("mean:",mean)
print ("Standard Deviation:",SD)
```

```python
dataset.head()
```

```python
# Step 3 — Plot the Histogram
```

```python
plt.hist(dataset["Supplier lead time (days)"],bins=10)
plt.xlabel = ["Supplier lead time (days)"]
plt.ylabel = ["Purchase orders"]
plt.title = ["Supplier lead time Analysis"]

plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(dataset["Supplier lead time (days)"],bins=10,kde=True)
plt.xlabel=("Supplier lead time (days)")
plt.ylabel=("Purchase orders")
plt.title=("Supplier lead time Analysis")

plt.show()
```
