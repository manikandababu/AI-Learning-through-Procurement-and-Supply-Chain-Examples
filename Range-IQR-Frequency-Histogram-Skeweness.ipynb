```python
# Frequency - Practical 1 — Count the Frequency
```

```python
from collections import Counter
lead_time = [4, 5, 4, 6, 5, 7, 5, 4, 6, 5]
frequency = Counter(lead_time)
frequency
```

```python
# Frequency - Frequency Distribution Table
```

```python
# Frequency Distribution Table - Step 1 — Sort the lead times
```

```python
sort_frequency = dict(sorted(frequency.items()))
sort_frequency
```

```python
# Frequency Distribution Table - Step 2 — Create a Frequency Table
```

```python
import pandas as pd
frequency_table = pd.DataFrame(list(sort_frequency.items()),columns=["Supplier Lead Time","Repeated Frequency"])
frequency_table
```

```python
# Frequency Distribution Table - Step 3 — Practical 3 — Relative Frequency (%)
```

```python
total_occurance = frequency_table["Repeated Frequency"].sum()
frequency_table["Frequency%"] = (frequency_table["Repeated Frequency"]/total_occurance*100)
frequency_table
```

```python
frequency_table["Cumulative Frequency"] = frequency_table["Repeated Frequency"].cumsum()
frequency_table["Cumulative Frequency%"] =  frequency_table["Frequency%"].cumsum()
frequency_table
```
